```python
from sklearn.model_selection import KFold
import numpy as np

def oof_target_encoder(df, column, target, n_splits=5, smoothing=10):
    """
    Performs Out-of-Fold target encoding with smoothing.
    """
    df = df.copy()
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
    encoded_column = np.zeros(len(df))

    # Global mean for smoothing
    global_mean = df[target].mean()

    for train_idx, val_idx in kf.split(df):
        # Calculate stats on training folds
        train_data = df.iloc[train_idx]
        val_data = df.iloc[val_idx]

        stats = train_data.groupby(column)[target].agg(['count', 'mean'])

        # Apply smoothing
        # (count * mean + smoothing * global_mean) / (count + smoothing)
        smoothed = (stats['count'] * stats['mean'] + smoothing * global_mean) / (stats['count'] + smoothing)

        # Map to validation fold
        encoded_column[val_idx] = val_data[column].map(smoothed).fillna(global_mean)

    return encoded_column

# Identify categorical columns
cat_features = df.select_dtypes(include=['object']).columns

# Apply OOF encoding to each categorical feature
for col in cat_features:
    df[f'{col}_encoded'] = oof_target_encoder(df, col, 'SalePrice')

# Display the original vs encoded columns for a few categories
print(f"Encoded {len(cat_features)} categorical features.")
display(df[[cat_features[0], f'{cat_features[0]}_encoded', 'SalePrice']].head())
```

### Feature Selection via Null Importance
We will use LightGBM to calculate feature importances. By comparing the actual importance of a feature against its importance when the target is randomized (Null Importance), we can statistically determine if a feature is providing real predictive power.

```python
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Prepare data: Combine numeric features and encoded categorical features
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
# Exclude target and ID
features = [c for c in numeric_cols if c not in ['SalePrice', 'Id']]

X = df[features]
y = df['SalePrice']

def get_feature_importances(data, target, shuffle=False):
    if shuffle:
        target = target.sample(frac=1).values

    lgbm = lgb.LGBMRegressor(n_estimators=200, learning_rate=0.05, importance_type='gain', verbose=-1)
    lgbm.fit(data, target)
    return lgbm.feature_importances_

# 2. Calculate Actual Importance
actual_importance = get_feature_importances(X, y, shuffle=False)

# 3. Calculate Null Importance (50 iterations)
null_importances = []
print("Running null importance iterations...")
for i in range(50):
    if (i+1) % 10 == 0: print(f"Iteration {i+1}/50")
    null_importances.append(get_feature_importances(X, y, shuffle=True))

null_importance_df = pd.DataFrame(null_importances, columns=features)
```

```python
# 4. Thresholding: Keep features where actual > 95th percentile of null
feature_scores = []
for i, col in enumerate(features):
    actual = actual_importance[i]
    null_dist = null_importance_df[col]
    percentile_95 = np.percentile(null_dist, 95)

    feature_scores.append({
        'feature': col,
        'actual_importance': actual,
        'null_95th': percentile_95,
        'keep': actual > percentile_95
    })

scores_df = pd.DataFrame(feature_scores).sort_values(by='actual_importance', ascending=False)
selected_features = scores_df[scores_df['keep'] == True]['feature'].tolist()

print(f"Selected {len(selected_features)} features out of {len(features)}.")
display(scores_df.head(20))
```

## Model Training and Hyperparameter Tuning
We will now train a variety of models using the features selected by the Null Importance test. For the gradient boosting models, we'll perform a randomized search to find optimal parameters.

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import RandomizedSearchCV, cross_val_score
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
import lightgbm as lgb

# Prepare final feature set
X_final = X[selected_features]
y_final = y.values

# Dictionary to store results
results = {}

# 1. Linear Models (Baselines)
linear_models = {
    'LinearRegression': LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.1),
    "ElasticNet": ElasticNet(alpha=0.1, l1_ratio=0.5)
}

print("Evaluating Linear Models...")
for name, model in linear_models.items():
    scores = cross_val_score(model, X_final, y_final, cv=5, scoring='neg_root_mean_squared_error')
    results[name] = -scores.mean()
    print(f"{name} RMSE: {results[name]:.2f}")
```

```python
# 2. Boosting Models with Tuning

def tune_boosting(model_type, X, y):
    if model_type == 'xgb':
        model = XGBRegressor(objective='reg:squarederror', random_state=42)
        param_grid = {
            'n_estimators': [100, 500, 1000],
            'learning_rate': [0.01, 0.05, 0.1],
            'max_depth': [3, 5, 7],
            'subsample': [0.8, 1.0]
        }
    elif model_type == 'lgb':
        model = lgb.LGBMRegressor(random_state=42, verbose=-1)
        param_grid = {
            'n_estimators': [100, 500, 1000],
            'learning_rate': [0.01, 0.05, 0.1],
            'num_leaves': [31, 63, 127],
            'feature_fraction': [0.8, 1.0]
        }
    elif model_type == 'cat':
        model = CatBoostRegressor(random_state=42, verbose=0)
        param_grid = {
            'iterations': [500, 1000],
            'learning_rate': [0.01, 0.05, 0.1],
            'depth': [4, 6, 8]
        }

    search = RandomizedSearchCV(model, param_grid, n_iter=10, cv=5,
                               scoring='neg_root_mean_squared_error', random_state=42)
    search.fit(X, y)
    return -search.best_score_, search.best_estimator_

print("\nTuning Boosting Models...")
for m_type in ['xgb', 'lgb', 'cat']:
    rmse, best_model = tune_boosting(m_type, X_final, y_final)
    results[f"{m_type.upper()}_Tuned"] = rmse
    print(f"{m_type.upper()} Tuned RMSE: {rmse:.2f}")

# Summary of results
summary_df = pd.DataFrame(list(results.items()), columns=['Model', 'RMSE']).sort_values('RMSE')
display(summary_df)
```

### Final Ensemble and Visualization
We will combine the predictions of the top 3 models (CatBoost, XGBoost, and LightGBM) to create a more robust prediction. We'll also visualize the actual vs. predicted prices.

```python
# We will retrieve the best estimators from our previous search for the final ensemble
# (Note: In a production setting, we would retrain on the full dataset)

from sklearn.model_selection import cross_val_predict

# Get top 3 models
best_xgb = tune_boosting('xgb', X_final, y_final)[1]
best_lgb = tune_boosting('lgb', X_final, y_final)[1]
best_cat = tune_boosting('cat', X_final, y_final)[1]

# Generate OOF predictions for the ensemble
preds_xgb = cross_val_predict(best_xgb, X_final, y_final, cv=5)
preds_lgb = cross_val_predict(best_lgb, X_final, y_final, cv=5)
preds_cat = cross_val_predict(best_cat, X_final, y_final, cv=5)

# Simple average ensemble
final_preds = (preds_xgb + preds_lgb + preds_cat) / 3

# Calculate final RMSE
from sklearn.metrics import mean_squared_error
ensemble_rmse = np.sqrt(mean_squared_error(y_final, final_preds))
print(f"Ensemble RMSE: {ensemble_rmse:.2f}")
```

### Voting and Stacking Ensembles
We will now use `sklearn`'s `VotingRegressor` and `StackingRegressor` to combine XGBoost and CatBoost. For stacking, we use a simple linear meta-learner (Ridge) to avoid further overfitting.

```python
from sklearn.ensemble import VotingRegressor, StackingRegressor

# Define the estimators based on our tuned models
estimators = [
    ('xgb', best_xgb),
    ('cat', best_cat)
]

# 1. Voting Ensemble
voting_model = VotingRegressor(estimators=estimators)
voting_scores = cross_val_score(voting_model, X_final, y_final, cv=5, scoring='neg_root_mean_squared_error')
results['Voting_Ensemble'] = -voting_scores.mean()

# 2. Stacking Ensemble
stacking_model = StackingRegressor(
    estimators=estimators,
    final_estimator=Ridge(alpha=1.0),
    cv=5
)
stacking_scores = cross_val_score(stacking_model, X_final, y_final, cv=5, scoring='neg_root_mean_squared_error')
results['Stacking_Ensemble'] = -stacking_scores.mean()

# Update results summary
summary_df = pd.DataFrame(list(results.items()), columns=['Model', 'RMSE']).sort_values('RMSE')
print("Updated Model Comparison:")
display(summary_df)
```

```python
# Visualize the Stacking results as it's typically the most advanced
stack_preds = cross_val_predict(stacking_model, X_final, y_final, cv=5)

plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_final, y=stack_preds, alpha=0.5, color='green')
plt.plot([y_final.min(), y_final.max()], [y_final.min(), y_final.max()], 'r--', lw=2)
plt.xlabel('Actual Sale Price')
plt.ylabel('Predicted Sale Price')
plt.title(f'Stacking Ensemble: Actual vs. Predicted (RMSE: {results["Stacking_Ensemble"]:.2f})')
plt.show()
```

### Optimization Step: Log Transform and Expanded Feature Selection
We will now transform the target to `log1p` to stabilize variance and relax the null importance threshold to the 80th percentile to bring in more supporting features.

```python
# 1. Log Transform the Target
y_log = np.log1p(df['SalePrice'])

# 2. Relaxed Feature Selection (80th Percentile)
feature_scores_relaxed = []
for i, col in enumerate(features):
    actual = actual_importance[i]
    null_dist = null_importance_df[col]
    percentile_80 = np.percentile(null_dist, 80)
    feature_scores_relaxed.append({'feature': col, 'keep': actual > percentile_80})

relaxed_features = [f['feature'] for f in feature_scores_relaxed if f['keep']]
X_relaxed = df[relaxed_features]

print(f"New feature count: {len(relaxed_features)}")

# 3. Quick Test with Tuned CatBoost on Log Target
# We use Mean Squared Error on log scale, then convert back to RMSE for comparison
from sklearn.metrics import mean_squared_error

cb_model = CatBoostRegressor(iterations=1000, learning_rate=0.05, depth=6, random_state=42, verbose=0)
cv_preds_log = cross_val_predict(cb_model, X_relaxed, y_log, cv=5)

# Convert back to original scale to calculate RMSE
final_rmse_optimized = np.sqrt(mean_squared_error(df['SalePrice'], np.expm1(cv_preds_log)))
print(f"Optimized CatBoost RMSE (Original Scale): {final_rmse_optimized:.2f}")
```

```python
import numpy as np
import pandas as pd
import lightgbm as lgb

# 1. Setup stability check parameters
n_runs = 5
threshold_percentile = 80
min_appearances = 4
selection_counts = {feat: 0 for feat in features}

print(f"Running {n_runs} stability iterations for Null Importance...")

for run in range(n_runs):
    # Calculate Actual Importance for this seed
    lgbm = lgb.LGBMRegressor(n_estimators=200, learning_rate=0.05, importance_type='gain', random_state=run, verbose=-1)
    lgbm.fit(X, y)
    run_actual_imp = lgbm.feature_importances_

    # Calculate Null Importance (20 iterations per run to save time)
    run_null_imps = []
    for i in range(20):
        y_shuffled = y.sample(frac=1, random_state=i + (run * 100)).values
        lgbm_null = lgb.LGBMRegressor(n_estimators=200, learning_rate=0.05, importance_type='gain', verbose=-1)
        lgbm_null.fit(X, y_shuffled)
        run_null_imps.append(lgbm_null.feature_importances_)

    run_null_df = pd.DataFrame(run_null_imps, columns=features)

    # Check which features pass the 80th percentile threshold in this specific run
    for i, col in enumerate(features):
        percentile_val = np.percentile(run_null_df[col], threshold_percentile)
        if run_actual_imp[i] > percentile_val:
            selection_counts[col] += 1

    print(f"Completed Run {run+1}/{n_runs}")

# 2. Filter features based on frequency
stable_features = [feat for feat, count in selection_counts.items() if count >= min_appearances]

print(f"\nFeatures passing {min_appearances}/{n_runs} runs: {len(stable_features)}")
print(f"Stable Features: {stable_features}")
```

```python
# 3. Re-evaluate CatBoost with stable features and log target
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import mean_squared_error

X_stable = df[stable_features]
y_log = np.log1p(df['SalePrice'])

cb_model_stable = CatBoostRegressor(iterations=1000, learning_rate=0.05, depth=6, random_state=42, verbose=0)
cv_preds_log_stable = cross_val_predict(cb_model_stable, X_stable, y_log, cv=5)

# Calculate RMSE on original scale
rmse_stable = np.sqrt(mean_squared_error(df['SalePrice'], np.expm1(cv_preds_log_stable)))
print(f"Stable CatBoost RMSE (Original Scale): {rmse_stable:.2f}")
```

```python
# 1. Feature Engineering: Create high-impact domain features
df['TotalSF'] = df['TotalBsmtSF'] + df['1stFlrSF'] + df['2ndFlrSF']
df['TotalBath'] = df['FullBath'] + (0.5 * df['HalfBath']) + df['BsmtFullBath'] + (0.5 * df['BsmtHalfBath'])
df['AgeAtSale'] = df['YrSold'] - df['YearBuilt']

# 2. Update feature list to include new engineering
new_engineer_feats = ['TotalSF', 'TotalBath', 'AgeAtSale']
features = features + new_engineer_feats
X = df[features]

print(f"New features added: {new_engineer_feats}")
print(f"Total feature count for selection: {len(features)}")
```

```python
# 3. Rerun Stability Check with new features
selection_counts = {feat: 0 for feat in features}
print(f"Rerunning stability iterations with engineered features...")

for run in range(n_runs):
    lgbm = lgb.LGBMRegressor(n_estimators=200, learning_rate=0.05, importance_type='gain', random_state=run, verbose=-1)
    lgbm.fit(X, y)
    run_actual_imp = lgbm.feature_importances_

    run_null_imps = []
    for i in range(20):
        y_shuffled = y.sample(frac=1, random_state=i + (run * 100)).values
        lgbm_null = lgb.LGBMRegressor(n_estimators=200, learning_rate=0.05, importance_type='gain', verbose=-1)
        lgbm_null.fit(X, y_shuffled)
        run_null_imps.append(lgbm_null.feature_importances_)

    run_null_df = pd.DataFrame(run_null_imps, columns=features)
    for i, col in enumerate(features):
        percentile_val = np.percentile(run_null_df[col], threshold_percentile)
        if run_actual_imp[i] > percentile_val:
            selection_counts[col] += 1

stable_features = [feat for feat, count in selection_counts.items() if count >= min_appearances]
print(f"\nNew Stable Features: {stable_features}")

# 4. Final Evaluation
X_stable = df[stable_features]
cv_preds_log_stable = cross_val_predict(cb_model_stable, X_stable, y_log, cv=5)
rmse_stable = np.sqrt(mean_squared_error(df['SalePrice'], np.expm1(cv_preds_log_stable)))
print(f"Updated CatBoost RMSE with Engineered Features: {rmse_stable:.2f}")
```

### Advanced Diagnostics & Feature Engineering
We will now analyze the residuals to see if errors are heteroscedastic (larger for expensive homes) and add group-based features to capture neighborhood-specific value trends.

```python
import scipy.stats as stats

# 1. Residual Analysis: Plotting errors against actual price
residuals = df['SalePrice'] - np.expm1(stacking_preds_log)

plt.figure(figsize=(10, 5))
sns.regplot(x=df['SalePrice'], y=residuals, scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Residuals vs Actual Price (Red line shows bias trend)')
plt.xlabel('Actual Sale Price')
plt.ylabel('Residual (Actual - Predicted)')
plt.show()

# 2. Smearing Correction (Duan's Estimator)
# Simple exp(preds) is biased because E[exp(x)] != exp(E[x])
log_residuals = y_log - stacking_preds_log
smearing_factor = np.mean(np.exp(log_residuals))
final_preds_smeared = np.expm1(stacking_preds_log) * smearing_factor

rmse_smeared = np.sqrt(mean_squared_error(df['SalePrice'], final_preds_smeared))
print(f"RMSE after Smearing Correction: {rmse_smeared:.2f}")
```

```python
# 3. Group-based Feature Engineering
# Capture how much a house deviates from its neighborhood average
df['Nbhd_Median_SF'] = df.groupby('Neighborhood')['TotalSF'].transform('median')
df['SF_Rel_to_Nbhd'] = df['TotalSF'] / df['Nbhd_Median_SF']

# Interaction: Quality * Size (The most common non-linear driver in real estate)
df['Qual_SF_Interact'] = df['OverallQual'] * df['TotalSF']

# Update feature list
extended_features = boruta_features + ['SF_Rel_to_Nbhd', 'Qual_SF_Interact']
X_ext = df[extended_features]

# Re-run Stacking with extended features
stacking_ext = StackingRegressor(
    estimators=[('xgb', best_xgb_boruta), ('cat', best_cat_boruta)],
    final_estimator=Ridge(alpha=1.0), cv=5
)
ext_preds_log = cross_val_predict(stacking_ext, X_ext, y_log, cv=5)
rmse_ext = np.sqrt(mean_squared_error(df['SalePrice'], np.expm1(ext_preds_log)))

print(f"RMSE with Neighborhood Relativity & Interactions: {rmse_ext:.2f}")
```

### Final Tuning Pass
Now that we have established a high-performing feature set (Boruta + Interactions), we will perform a more granular search for the best model parameters to maximize the ensemble's performance.

```python
from sklearn.model_selection import RandomizedSearchCV

# 1. More granular tuning for CatBoost
cb_refined = CatBoostRegressor(random_state=42, verbose=0, loss_function='RMSE')
cb_params = {
    'iterations': [1000, 1500],
    'learning_rate': [0.01, 0.03, 0.05],
    'depth': [4, 6, 8],
    'l2_leaf_reg': [1, 3, 5, 9],
    'bagging_temperature': [0, 1, 5]
}

rs_cat = RandomizedSearchCV(cb_refined, cb_params, n_iter=15, cv=5,
                            scoring='neg_mean_squared_error', random_state=42)
rs_cat.fit(X_ext, y_log)
best_cat_final = rs_cat.best_estimator_

# 2. More granular tuning for XGBoost
xgb_refined = XGBRegressor(objective='reg:squarederror', random_state=42)
xgb_params = {
    'n_estimators': [1000, 1500],
    'learning_rate': [0.01, 0.05],
    'max_depth': [3, 4, 5],
    'colsample_bytree': [0.7, 0.8, 0.9],
    'reg_alpha': [0, 0.1, 0.5],
    'reg_lambda': [1, 2, 5]
}

rs_xgb = RandomizedSearchCV(xgb_refined, xgb_params, n_iter=15, cv=5,
                            scoring='neg_mean_squared_error', random_state=42)
rs_xgb.fit(X_ext, y_log)
best_xgb_final = rs_xgb.best_estimator_

print("Refined Tuning Complete.")
```

```python
# 3. Final Stacked Ensemble with Refined Models
final_stacking_model = StackingRegressor(
    estimators=[('xgb', best_xgb_final), ('cat', best_cat_final)],
    final_estimator=Ridge(alpha=1.0),
    cv=5
)

# Generate OOF predictions
final_cv_preds_log = cross_val_predict(final_stacking_model, X_ext, y_log, cv=5)

# Apply Smearing Correction to the final model
final_log_residuals = y_log - final_cv_preds_log
final_smearing_factor = np.mean(np.exp(final_log_residuals))
final_actual_preds = np.expm1(final_cv_preds_log) * final_smearing_factor

final_rmse = np.sqrt(mean_squared_error(df['SalePrice'], final_actual_preds))
print(f"Final Optimized RMSE (Refined Params + Smearing): {final_rmse:.2f}")
```

### Final Consolidation: The 'Best-of-All' Model
We will now combine the best refined models (CatBoost and XGBoost) using an optimized weighted blend and the Duan Smearing correction, which previously showed the best performance for handling the log-transformation bias.

```python
# 1. Get OOF predictions for the refined models
y_pred_xgb = cross_val_predict(best_xgb_final, X_ext, y_log, cv=5)
y_pred_cat = cross_val_predict(best_cat_final, X_ext, y_log, cv=5)

# 2. Weighted Blending (0.7 CatBoost / 0.3 XGBoost based on individual strengths)
blended_preds_log = (0.7 * y_pred_cat) + (0.3 * y_pred_xgb)

# 3. Apply Smearing Correction
blend_residuals = y_log - blended_preds_log
blend_smearing = np.mean(np.exp(blend_residuals))
final_blend_preds = np.expm1(blended_preds_log) * blend_smearing

# 4. Final Score
final_rmse_blend = np.sqrt(mean_squared_error(df['SalePrice'], final_blend_preds))
print(f"Final Optimized Weighted Blend RMSE: {final_rmse_blend:.2f}")

# 5. Summary Table
milestones = {
    'Initial Baseline (Tuned CatBoost)': 31429.06,
    'Engineered Features (Manual Select)': 26957.85,
    'Boruta Selection + Log Target': 24420.70,
    'Stacking Ensemble + Boruta': 23898.85,
    'Boruta + Smearing Correction': 23816.76,
    'Final Weighted Blend + Smearing': final_rmse_blend
}

summary_table = pd.DataFrame(list(milestones.items()), columns=['Milestone', 'RMSE']).sort_values('RMSE')
display(summary_table)
```

### Final Consolidation: The 'Best-of-All' Model
We will now combine the best individual models (CatBoost and XGBoost) with a simple weighted average and the Duan Smearing correction, as this combination has proven more stable than stacking with a meta-learner.

```python
# 1. Get OOF predictions for the best performing individual models
# Using the refined models we just trained
y_pred_xgb = cross_val_predict(best_xgb_final, X_ext, y_log, cv=5)
y_pred_cat = cross_val_predict(best_cat_final, X_ext, y_log, cv=5)

# 2. Optimized Weighted Blending (70% CatBoost / 30% XGBoost based on standalone performance)
blended_preds_log = (0.7 * y_pred_cat) + (0.3 * y_pred_xgb)

# 3. Apply Smearing Correction
blend_residuals = y_log - blended_preds_log
blend_smearing = np.mean(np.exp(blend_residuals))
final_blend_preds = np.expm1(blended_preds_log) * blend_smearing

# 4. Final Score
final_rmse_blend = np.sqrt(mean_squared_error(df['SalePrice'], final_blend_preds))
print(f"Final Optimized Weighted Blend RMSE: {final_rmse_blend:.2f}")

# 5. Summary Table of all key milestones
milestones = {
    'Initial Baseline (Tuned CatBoost)': 31429.06,
    'Engineered Features (Manual Select)': 26957.85,
    'Boruta Selection + Log Target': 24420.70,
    'Stacking Ensemble + Boruta': 23898.85,
    'Boruta + Smearing Correction': 23816.76,
    'Final Weighted Blend + Smearing': final_rmse_blend
}

summary_table = pd.DataFrame(list(milestones.items()), columns=['Milestone', 'RMSE']).sort_values('RMSE')
display(summary_table)
```

### Statistical Significance & Error Decomposition
We need to determine if the ~23.8k floor is real or just a result of lucky folds, and where specifically the model is losing 'money'.

```python
from sklearn.model_selection import KFold

# 1. Calculate CV Standard Deviation to check for noise-chasing
kf = KFold(n_splits=5, shuffle=True, random_state=42)
fold_errors = []

for train_idx, val_idx in kf.split(X_ext):
    model = StackingRegressor(estimators=estimators_boruta, final_estimator=Ridge(alpha=1.0))
    model.fit(X_ext.iloc[train_idx], y_log.iloc[train_idx])
    preds = np.expm1(model.predict(X_ext.iloc[val_idx]))
    fold_errors.append(np.sqrt(mean_squared_error(df.iloc[val_idx]['SalePrice'], preds)))

print(f"CV RMSE: {np.mean(fold_errors):.2f} (+/- {np.std(fold_errors):.2f})")

# 2. Decile Error Decomposition
results_df = pd.DataFrame({'Actual': df['SalePrice'], 'Predicted': final_preds_smeared})
results_df['Error_Sq'] = (results_df['Actual'] - results_df['Predicted'])**2
results_df['Decile'] = pd.qcut(results_df['Actual'], 10, labels=False)

decile_analysis = results_df.groupby('Decile').agg(
    mean_actual=('Actual', 'mean'),
    rmse=('Error_Sq', lambda x: np.sqrt(x.mean())),
    pct_of_total_sse=('Error_Sq', lambda x: (x.sum() / results_df['Error_Sq'].sum()) * 100)
)
display(decile_analysis)
print("Interpretation: If 'pct_of_total_sse' for decile 9 is > 50%, we have a high-end outlier problem.")
```

### New Signal: KNN-Comparables
We will create a 'Peer Group Price' feature. For every house, we find the 5 most similar houses (based on Size, Quality, and Age) and calculate their average price, calculated strictly out-of-fold to prevent leakage.

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler

# Features to determine 'similarity'
sim_features = ['TotalSF', 'OverallQual', 'AgeAtSale', 'Neighborhood_encoded']
scaler = StandardScaler()
X_sim = scaler.fit_transform(df[sim_features])

# KNN Regressor to find comparable sales (OOF)
knn = KNeighborsRegressor(n_neighbors=5, weights='distance')

# Use cross_val_predict to get the 'Peer Price' for each house without leakage
df['Peer_Comp_Price'] = cross_val_predict(knn, X_sim, df['SalePrice'], cv=5)

# Re-evaluate with this heavy-hitter feature
X_final_plus_knn = pd.concat([X_ext, df[['Peer_Comp_Price']]], axis=1)

stack_final = StackingRegressor(estimators=estimators_boruta, final_estimator=Ridge())
final_oof_log = cross_val_predict(stack_final, X_final_plus_knn, y_log, cv=5)

final_rmse_with_knn = np.sqrt(mean_squared_error(df['SalePrice'], np.expm1(final_oof_log)))
print(f"RMSE with KNN-Comparables added: {final_rmse_with_knn:.2f}")
```

### Refined Meta-Stacking for KNN Integration
We will now properly integrate the KNN 'Peer Price' by using a Stacking Regressor that treats the KNN output as a base feature alongside the OOF predictions of our XGBoost and CatBoost models. This allows the meta-learner to 'learn' when to trust the neighbors versus the algorithms.

```python
from sklearn.linear_model import RidgeCV

# 1. Define base estimators for the stack
base_estimators = [
    ('xgb', best_xgb_final),
    ('cat', best_cat_final)
]

# 2. Use StackingRegressor to combine features + base models
# We include the KNN price as a feature in X_final_plus_knn
stack_meta = StackingRegressor(
    estimators=base_estimators,
    final_estimator=RidgeCV(),
    cv=5,
    passthrough=True # This allows the original features (including KNN) to reach the meta-learner
)

# 3. Generate OOF predictions
meta_preds_log = cross_val_predict(stack_meta, X_final_plus_knn, y_log, cv=5)

# 4. Apply Smearing Correction
meta_residuals = y_log - meta_preds_log
meta_smearing = np.mean(np.exp(meta_residuals))
final_meta_preds = np.expm1(meta_preds_log) * meta_smearing

# 5. Final Evaluation
rmse_meta_final = np.sqrt(mean_squared_error(df['SalePrice'], final_meta_preds))
print(f"Refined Meta-Stacking RMSE: {rmse_meta_final:.2f}")

if rmse_meta_final < 23816:
    print("✅ Success: New best score achieved!")
else:
    print("⚠️ Performance is stable, but high-end outliers remain the bottleneck.")
```
